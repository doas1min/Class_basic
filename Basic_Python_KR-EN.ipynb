**(3) 배열의 각 원소 (인덱스) (Each element of an array — index)**

*   첫번째 학생 점수 (1st student's score): 92
*   두번째 학생 점수 (2nd student's score): 83
*   세번째 학생 점수 (3rd student's score): 87

**(4) 슬라이싱: 배열의 원소 중 일정한 부분만 가져오기 (Slicing: taking only a certain part of the elements of an array)**

list[a:b] : 인덱스 a 번부터 (b-1)번까지 가져오기 (take the elements from index a up to index b-1)

````python
list = [1,3,5,7,9,11,13,15]
print(list[2:6])  # 인덱스 2번부터 5번까지 (index 2 to index 5)
print(list[2:])   # 인덱스 2번부터 마지막까지 (index 2 to the end)
print(list[:5])   # 인덱스 0번부터 4번까지 (index 0 to index 4)
````

# **원주율 구하기 (Calculating pi)**

pi2.png



1.   rand() 함수 사용 (Use the rand() function)
2.   입력값을 시행횟수, 출력값을 계산된 $\pi$ 값으로 하는 함수 만들기 (Write a function whose input is the number of trials and whose output is the estimated $\pi$)
3. for문, if문 사용 (Use for statements and if statements)



````python
import numpy as np

def cal_pi(number):
    num_in = 0  # 원 안에 들어간 점의 개수 (Number of points that fall inside the circle)
    for i in range(0,number):
        x = np.random.rand()  # 0~1 사이의 난수 (A random number between 0 and 1)
        y = np.random.rand()
        r = (x**2+y**2)**0.5  # 원점으로부터의 거리 (Distance from the origin)
        if r <= 1.0:
            num_in = num_in + 1
    expected_pi = float(num_in/number)*4.0  # 넓이 비율로부터 pi 추정 (Estimate pi from the ratio of areas)
    return expected_pi
````

````python
for i in range(0,8):
  n = 10**(i+1) # 0 ~ 4
  pi = cal_pi(n)
  print ("시행횟수 (number of trials) = 10^",(i+1),":",pi)
````

## **그래프 그리기 (Plotting graphs): matplot 라이브러리 (the matplotlib library)**

matplot.png

````python
import matplotlib.pyplot as plt
import numpy as np

# x 값 범위 설정 (Set the range of x values)
x = np.linspace(-5, 5, 400)  # -10부터 10까지 400개의 점 (400 points from -10 to 10)

# y = x^2 계산 (Compute y = x^2)
y = x**2

# 그래프 그리기 (Draw the graph)
plt.plot(x, y, 'b-', label='y = x^2')

# 그래프 제목과 축 이름 (Graph title and axis labels)
plt.title('Graph of y = x^2')
plt.xlabel('x')
plt.ylabel('y')

# 범례 표시 (Show the legend)
plt.legend()

# 격자 표시 (Show the grid)
plt.grid(True)

# 그래프 출력 (Display the graph)
plt.show()
````
